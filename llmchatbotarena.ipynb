```python
#nettoyage des données
from spellchecker import SpellChecker
import re
import emoji

def nettoyerlesdonnes(texte):
  texte = re.sub(r'<[^>]+>','',texte) # supprime les balises html
  texte= emoji.replace_emoji(texte, replace='')
  return texte
spell = SpellChecker(language = 'en')
def corrigerlesmots(texte):
  mots= texte.split()
  mots_corrigés= [spell.correction(mot)if spell.correction(mot) is not None else mot for mot in mots]
  return ''.join(mots_corrigés)
```
